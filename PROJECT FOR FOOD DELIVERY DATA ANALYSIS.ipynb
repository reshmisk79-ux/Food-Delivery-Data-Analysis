 Handle Missing Values

```python
print(food_clean.isnull().sum())
```

Correct Date Columns

```python
date_columns = [
    col for col in food_clean.columns
    if "date" in col.lower()
]

print(date_columns)
```

```python
print(food_clean[date_columns].dtypes)
```

Create Date-Based Columns

```python
import pandas as pd

if len(date_columns) > 0:
    date_col = date_columns[0]

    # Convert the column into datetime format
    food_clean[date_col] = pd.to_datetime(
        food_clean[date_col],
        errors="coerce"
    )

    # Extract date information
    food_clean["Order_Day"] = food_clean[date_col].dt.day
    food_clean["Order_Month"] = food_clean[date_col].dt.month
    food_clean["Order_Year"] = food_clean[date_col].dt.year

    print(food_clean.head())
```

Create Total Amount of Each item

```python
print(food_clean.columns.tolist())
```

```python
if "items_count" in food_clean.columns and "order_total" in food_clean.columns:
    food_clean["Calculated_Total"] = (
        food_clean["items_count"] * food_clean["order_total"]
    )
```

```python
food_clean[["items_count", "order_total", "Calculated_Total"]].head()
```

Create Delivery Time Categories

```python
if "actual_delivery_time_minutes" in food_clean.columns:

    def delivery_partner_rating(time):
        if time <= 20:
            return "Fast"
        elif time <= 40:
            return "Medium"
        else:
            return "Slow"

    food_clean["delivery_partner_rating"] = (
        food_clean["actual_delivery_time_minutes"].apply(delivery_partner_rating)
    )
```

```python
food_clean[["actual_delivery_time_minutes", "delivery_partner_rating"]].head()
```

Create Order Value Categories

```python
if "order_total" in food_clean.columns:

    def subtotal(amount):
        if amount < 500:
            return "Low"
        elif amount < 1500:
            return "Medium"
        else:
            return "High"

    food_clean["subtotal"] = (
        food_clean["order_total"].apply(subtotal)
    )
```

```python
food_clean[["order_total", "subtotal"]].head()
```

Histogram

 Bar Chart

```python
col = food_clean.select_dtypes(include="object").columns[0]

food_clean[col].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Categories")
plt.xlabel(col)
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.show()
```

The bar chart created from food_clean is used to compare the number of orders/records across different categories.

The X-axis shows the different categories.
The Y-axis shows the number of orders.
A taller bar means that category has more orders.
A shorter bar means that category has fewer orders.
It helps us quickly identify the most and least frequently occurring categories.

A scatter plot is used to study the relationship between two numerical variables in the food delivery dataset.

The X-axis represents one numerical variable.
The Y-axis represents another numerical variable.
Each point represents one record or order.
Points moving upward together may indicate a positive relationship.
Points moving downward may indicate a negative relationship.
Randomly scattered points may indicate little or no relationship.
It can also help identify unusual values or outliers.
